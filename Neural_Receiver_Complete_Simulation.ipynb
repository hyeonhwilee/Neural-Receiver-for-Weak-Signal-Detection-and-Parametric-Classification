## 🔬 Step 5: 개별 신호 테스트 (선택사항)

학습된 모델로 특정 신호 테스트

```python
import numpy as np
import torch
from src.data import SignalGenerator
from src.data.signal_generator import SignalParams, ModulationType
from src.models import NeuralReceiver

# GPU 확인
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Device: {device}")

# 모델 로드
print("\n모델 로드 중...")
model = NeuralReceiver(
    input_channels=2,
    base_channels=64,
    num_blocks=4,
    num_classes=12,
    dropout=0.3
).to(device)

checkpoint = torch.load('experiments/neural_receiver/checkpoints/best_model.pt',
                       map_location=device)
model.load_state_dict(checkpoint['model_state_dict'])
model.eval()
print("✓ 모델 로드 완료")

# 테스트 신호 생성
signal_gen = SignalGenerator(seed=42)

# 여기를 변경하여 다른 변조 타입 테스트: 'AM', 'FM', '2FSK', 'CW Radar', 'LFM/Chirp' 등
test_modulation = 'FM'

test_params = SignalParams(
    center_freq=0.1,
    bandwidth=0.05,
    power=1.0,
    snr_db=-8.0,
    symbol_rate=0.05,
    modulation_type=test_modulation
)

_, test_signal = signal_gen.generate_signal(1024, test_params)

print("\n" + "="*70)
print("🧪 테스트 신호 정보")
print("="*70)
print(f"변조 타입:        {test_params.modulation_type}")
print(f"SNR:             {test_params.snr_db} dB")
print(f"Center Freq:     {test_params.center_freq:.4f}")
print(f"Bandwidth:       {test_params.bandwidth:.4f}")

# 추론
iq_tensor = torch.stack([
    torch.from_numpy(test_signal.real).float(),
    torch.from_numpy(test_signal.imag).float()
]).unsqueeze(0).to(device)

with torch.no_grad():
    predictions = model.predict(iq_tensor)

# 결과 출력
modulation_classes = ModulationType.get_class_names()

print("\n" + "="*70)
print("🔍 모델 추론 결과")
print("="*70)
print(f"신호 검출:        {'✅ 검출됨' if predictions['signal_detected'].item() else '❌ 검출 안됨'}")
print(f"검출 확률:        {predictions['detection_prob'].item():.4f}")
print(f"예측 변조 타입:   {modulation_classes[predictions['modulation_class'].item()]}")
print(f"분류 확신도:      {predictions['classification_confidence'].item():.4f}")

print(f"\n추정 파라미터:")
print(f"  Center Freq:    {predictions['parameters'][0, 0].item():.4f} (실제: {test_params.center_freq:.4f})")
print(f"  Bandwidth:      {predictions['parameters'][0, 1].item():.4f} (실제: {test_params.bandwidth:.4f})")
print(f"  SNR:            {predictions['parameters'][0, 2].item():.2f} dB (실제: {test_params.snr_db:.2f} dB)")
print(f"  Symbol Rate:    {predictions['parameters'][0, 3].item():.4f} (실제: {test_params.symbol_rate:.4f})")

print(f"\n오차:")
print(f"  Center Freq:    {abs(predictions['parameters'][0, 0].item() - test_params.center_freq):.4f}")
print(f"  Bandwidth:      {abs(predictions['parameters'][0, 1].item() - test_params.bandwidth):.4f}")
print(f"  SNR:            {abs(predictions['parameters'][0, 2].item() - test_params.snr_db):.2f} dB")
print(f"  Symbol Rate:    {abs(predictions['parameters'][0, 3].item() - test_params.symbol_rate):.4f}")
print("="*70)
```
